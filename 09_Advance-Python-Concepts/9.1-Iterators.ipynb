##### Iterators:
Iterators are advanced Python Concepts that allow for efficient looping and memory management. Iterators provide a way to access the elements of a collection sequentially without exposing its underlying representation.

```python
MY_LIST = [1, 2, 3, 4, 5, 6, 7]
for i in MY_LIST:
    print(i)
```

```python
type(MY_LIST)
```

```python
## Iterators and Iterables in Python
iterator = iter(MY_LIST)
print(type(iterator))
```

```python
iterator
```

```python
## Iterate through all the elements using the next() function
print(next(iterator))
```

```python
iterator = iter(MY_LIST)
```

```python
try:
    print(next(iterator))
except StopIteration:
    print("End of Iterator reached")
```

```python
my_string = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator))  # H
print(next(string_iterator))  # e
print(next(string_iterator))  # l
print(next(string_iterator))  # l
print(next(string_iterator))  # o
```
