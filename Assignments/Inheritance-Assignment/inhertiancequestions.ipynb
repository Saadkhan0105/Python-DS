### Assignment 1: Single Inheritance Basic

Create a base class named `Animal` with attributes `name` and `species`. Create a derived class named `Dog` that inherits from `Animal` and adds an attribute `breed`. Create an object of the `Dog` class and print its attributes.

```python
# Base class
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

# Derived class
class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name, species)  # Call the parent class constructor
        self.breed = breed

# Create an object of the Dog class
dog1 = Dog("Buddy", "Canine", "Golden Retriever")

# Print its attributes
print(f"Name: {dog1.name}")
print(f"Species: {dog1.species}")
print(f"Breed: {dog1.breed}")
```

### Assignment 2: Method Overriding in Single Inheritance

In the `Dog` class, override the `__str__` method to return a string representation of the object. Create an object of the class and print it.

```python
# Base class
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

# Derived class
class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name, species)
        self.breed = breed

    # Overriding the __str__ method
    def __str__(self):
        return f"Dog(Name: {self.name}, Species: {self.species}, Breed: {self.breed})"

# Create an object of Dog class
dog1 = Dog("Buddy", "Canine", "Golden Retriever")

# Print the object (this will call the overridden __str__ method)
print(dog1)
```

### Assignment 5: Method Resolution Order (MRO) in Multiple Inheritance

In the Athlete class, override the walk method to print a different message. Create an object of the class and call the walk method. Use the super() function to call the walk method of the Walker class.

```python
# Base class 1
class Walker:
    def walk(self):
        print("Walker: Walking at a steady pace.")

# Base class 2
class Runner:
    def run(self):
        print("Runner: Running fast and strong!")

# Derived class
class Athlete(Walker, Runner):
    # Overriding the walk method
    def walk(self):
        print("Athlete: Preparing to walk (overridden).")
        super().walk()  # Calls Walker's walk() because of MRO

# Create an object of Athlete class
athlete1 = Athlete()

# Call the walk method
athlete1.walk()

# Display Method Resolution Order
print("MRO:", [cls.__name__ for cls in Athlete.__mro__])
```

### Assignment 6: Multiple Inheritance with Additional Attributes

In the Athlete class, add an attribute training_hours and a method train that prints the training hours. Create an object of the class and call the method.

```python
# Base class 1
class Walker:
    def walk(self):
        print("Walker: Walking at a steady pace.")

# Base class 2
class Runner:
    def run(self):
        print("Runner: Running fast and strong!")

# Derived class
class Athlete(Walker, Runner):
    def __init__(self, training_hours):
        self.training_hours = training_hours  # Additional attribute

    def train(self):
        print(f"Athlete: Training for {self.training_hours} hour(s) today.")

# Create an object of the Athlete class
athlete1 = Athlete(training_hours=3)

# Call the train method
athlete1.train()

# Call inherited methods for verification
athlete1.walk()
athlete1.run()
```

### Assignment 7: Diamond Problem in Multiple Inheritance

Create a class named A with a method show that prints a message. Create two derived classes B and C that inherit from A and override the show method. Create a class D that inherits from both B and C. Create an object of the D class and call the show method. Observe the method resolution order.

```python
# Base class
class A:
    def show(self):
        print("Class A: show() method")

# Derived class 1
class B(A):
    def show(self):
        print("Class B: show() method")

# Derived class 2
class C(A):
    def show(self):
        print("Class C: show() method")

# Derived class D inherits from both B and C
class D(B, C):
    pass

# Create an object of class D
d = D()

# Call the show() method
d.show()

# Display the Method Resolution Order (MRO)
print("MRO:", [cls.__name__ for cls in D.__mro__])
```

### Assignment 9: Using super() in Multiple Inheritance

Create a class named Person with an attribute name. Create a class named Employee with an attribute employee_id. Create a derived class Manager that inherits from both Person and Employee. Use the super() function to initialize the attributes. Create an object of the Manager class and print its attributes.

```python
# Base class 1
class Person:
    def __init__(self, name, **kwargs):
        super().__init__(**kwargs)  # Allows cooperative multiple inheritance
        self.name = name

# Base class 2
class Employee:
    def __init__(self, employee_id, **kwargs):
        super().__init__(**kwargs)
        self.employee_id = employee_id

# Derived class inheriting from both Person and Employee
class Manager(Person, Employee):
    def __init__(self, name, employee_id):
        # Using cooperative super() call to initialize both parent classes
        super().__init__(name=name, employee_id=employee_id)

# Create an object of the Manager class
manager1 = Manager("Alice", "EMP102")

# Print attributes
print(f"Manager Name: {manager1.name}")
print(f"Employee ID: {manager1.employee_id}")

# Show the Method Resolution Order (MRO)
print("MRO:", [cls.__name__ for cls in Manager.__mro__])
```

### Assignment 10: Method Overriding and super()

Create a class named Vehicle with a method start that prints a starting message. Create a derived class Car that overrides the start method to print a different message. ### Assignment 12: Complex Multiple Inheritance

Create a class named Base1 with an attribute a. Create a class named Base2 with an attribute b. Create a class named Derived that inherits from both Base1 and Base2 and adds an attribute c. Initialize all attributes using the super() function. Create an object of the Derived class and print its attributes.

```python
# Base class 1
class Base1:
    def __init__(self, a, **kwargs):
        super().__init__(**kwargs)  
        self.a = a

# Base class 2
class Base2:
    def __init__(self, b, **kwargs):
        super().__init__(**kwargs)
        self.b = b

# Derived class inheriting from both Base1 and Base2
class Derived(Base1, Base2):
    def __init__(self, a, b, c):
        super().__init__(a=a, b=b)  
        self.c = c

# Create an object of the Derived class
obj = Derived(a=10, b=20, c=30)

# Print attributes
print(f"a = {obj.a}")
print(f"b = {obj.b}")
print(f"c = {obj.c}")

# Display the Method Resolution Order
print("MRO:", [cls.__name__ for cls in Derived.__mro__])
```

### Assignment 14: Polymorphism with Inheritance

Create a base class named Bird with a method speak. Create two derived classes Parrot and Penguin that override the speak method. Create a list of Bird objects and call the speak method on each object to demonstrate polymorphism.

```python
# Base class
class Bird:
    def speak(self):
        print("Bird: Chirp chirp!")

# Derived class 1
class Parrot(Bird):
    def speak(self):
        print("Parrot: Squawk! I can talk!")

# Derived class 2
class Penguin(Bird):
    def speak(self):
        print("Penguin: Honk! I prefer the cold!")

# Create a list of Bird objects (polymorphism in action)
birds = [Parrot(), Penguin(), Bird()]

# Call the speak() method on each object
for bird in birds:
    bird.speak()
```

### Assignment 15: Combining Single and Multiple Inheritance

Create a base class named Device with an attribute brand. Create a derived class Phone that inherits from Device and adds an attribute model. Create another base class Camera with an attribute resolution. Create a derived class Smartphone that inherits from both Phone and Camera. Create an object of the Smartphone class and print its attributes.

```python
# Base class 1
class Device:
    def __init__(self, brand, **kwargs):
        super().__init__(**kwargs)
        self.brand = brand

# Base class 2
class Camera:
    def __init__(self, resolution, **kwargs):
        super().__init__(**kwargs)
        self.resolution = resolution

# Derived class from Device
class Phone(Device):
    def __init__(self, brand, model, **kwargs):
        super().__init__(brand=brand, **kwargs)
        self.model = model

# Derived class from both Phone and Camera
class Smartphone(Phone, Camera):
    def __init__(self, brand, model, resolution):
        super().__init__(brand=brand, model=model, resolution=resolution)

# Create an object of Smartphone
smartphone1 = Smartphone("TechCorp", "X200", "108MP")

# Print attributes
print(f"Brand: {smartphone1.brand}")
print(f"Model: {smartphone1.model}")
print(f"Resolution: {smartphone1.resolution}")

# Show the Method Resolution Order (MRO)
print("MRO:", [cls.__name__ for cls in Smartphone.__mro__])
```
