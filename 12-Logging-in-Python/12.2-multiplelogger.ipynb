### Logging with Multiple Loggers

- We can create multiple loggers in Python and use them to log different types of messages.
- Each logger can have its own configuration and output destination.
- We can use the `logging.getLogger()` method to create a new logger.
- We can then use the `logger.debug()`, `logger.info()`, `logger.warning()`, `logger.error()`, and `logger.critical()` methods to log messages at different levels of severity.
- We can also use the `logger.exception()` method to log an exception that occurs in our code.
- We can use the `logging.basicConfig()` method to set the default configuration for all loggers.
- We can also use the `logging.shutdown()` method to clean up and release all resources used by the logging system.


```python
import logging

## Create and configure logger for module1
logger1 = logging.getLogger('module1')
logger1.setLevel(logging.DEBUG)

## Create and configure logger for module2
logger2 = logging.getLogger('module2')
logger2.setLevel(logging.WARNING)

## Cinfigure logging settings
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
## logging messages with different loggers
logger1.debug("This is debug message for module1")
logger2.warning("This is a warning message for module 2")
logger2.error("This is an error message")
```
