##### Data Manipulation and Analysis with Pandas

Data manipulation and analysis are key tasks in any data analysis project. Pandas provides a wiide range of functions for data manipulation and analysis, making it easier to clean, transform and extract inssights from data. 

```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
## fetcing the first 5 rows of the data
df.head(5)
```

```python
## fetcing the last 5 rows of the data
df.tail(5)
```

```python
df.dtypes
```

```python
## Handling missing data

df.isnull().any()
```

```python
df.isnull().sum()
```

```python
df_filled = df.fillna(0)
```

```python
### Filling missing values with the mean of the column

df['Sales_fillNA'] = df['Sales'].fillna(df['Sales'].mean())

print(df)
```

```python
df.dtypes
```

```python
## Renaming Columns
df.rename(columns={'Date': 'Sales Date'})
df.head()
```

```python
## change datatypes
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype('int')
df.head()
```

```python
df['New Value']=df['Value'].apply(lambda x:x*2)
df.head()
```

```python
## Data Aggregating And Grouping
df.head()
```

```python
grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Product','Region'])['Value'].sum()
print(grouped_sum)
```

```python
df.groupby(['Product','Region'])['Value'].mean()
```

```python
## aggregate multiple functions
groudped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
groudped_agg
```

```python
### Merging and joining Dataframes
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
## Merge Datafranme on the 'Key Columns'
pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```
