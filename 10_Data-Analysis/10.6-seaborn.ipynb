### Data Visualization With Seaborn

Seaborn is a Python Visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code.

```python
import seaborn as sns
```

```python
## Basic Plotting with Seaborn

tips = sns.load_dataset('tips')
print(tips)
```

```python
## create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Scatter Plot of Total Bill vs Tip")
plt.show()
```

```python
## Line Plot

sns.lineplot(x='size', y='total_bill', data=tips)
plt.title("Line Plot of Size vs Total Bill")
plt.show()
```

```python
## Categorical Plot

## Bar Plot
sns.barplot(x='day', y='total_bill', data=tips)
plt.title("Bar Plot of Total Bill by Day")
plt.show()
```

```python
## Box Plot

sns.boxplot(x='day', y='total_bill', data=tips)
plt.title("Box Plot of Total Bill by Day")
plt.show()
```

```python
## Violin Plot
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title("Violin Plot of Total Bill by Day")
plt.show()
```

```python
## Histogram
sns.histplot(tips['total_bill'], bins=10, kde=True)
plt.title("Histogram of Total Bill")
plt.show()
```

```python
## KDE Plot
sns.kdeplot(tips['total_bill'], fill=True)
plt.title("KDE Plot of Total Bill")
plt.show()
```

```python
## Pair Plot
sns.pairplot(tips)
plt.show()
```

```python
## Heatmap
corr = tips[['total_bill', 'tip', 'size']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap of Correlation Matrix")
plt.show()
```

```python
import pandas as pd
sales_df=pd.read_csv('sales_data.csv')
sales_df.head()
```

```python
## Plot total sales by product category

plt.figure(figsize=(10,6))
sns.barplot(x='Product Category', y='Total Revenue', data=sales_df, estimator=sum)
plt.title("Total Sales by Product Category")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

```python
## Plot total sales by product region

plt.figure(figsize=(10,6))
sns.barplot(x='Region', y='Total Revenue', data=sales_df, estimator=sum)
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```
